```ocaml
let rec rimuovi n lis =
if (List.length lis <= n) then [] else if n = 0 then lis else rimuovi (n-1) (List.tl lis);;
```

```ocaml
let rec rimuovi n lis = 
match (n,lis) with 
|(0,x) -> x
|(_, []) -> []
|(m, y::lis') -> rimuovi (m-1) lis';;
```

```ocaml
rimuovi 3 [1;2];;
```
